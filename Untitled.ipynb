```python
"""
Código para testar acesso à API SOAP do SIAFI/Tesouro Nacional
Requisitos: pip install requests zeep lxml

IMPORTANTE: A API SOAP do SIAFI requer:
1. Certificado digital A1 de equipamento (ICP-Brasil)
2. Cadastro prévio do sistema no Tesouro Nacional
3. Usuários cadastrados no SIAFI com perfis adequados
"""

import requests
from zeep import Client
from zeep.transports import Transport
from requests.auth import HTTPBasicAuth
import warnings
warnings.filterwarnings('ignore')

# ============================================
# CONFIGURAÇÕES
# ============================================

# Suas credenciais do Tesouro Gerencial
USUARIO = "macieiracesar@gmail.com"  # Ex: macieiracesar@gmail.com
SENHA = "Atletico@13"

# Ano do SIAFI (ano corrente)
ANO_SIAFI = "2026"

# Ambiente: 'homologacao' ou 'producao'
AMBIENTE = "homologacao"  # Comece sempre com homologação

# ============================================
# URLs das APIs SOAP do SIAFI
# ============================================

# URLs base dos WSDLs (substitua <ano> pelo ano atual)
WSDL_URLS = {
    "homologacao": {
        "contas_pagar_receber": f"https://homextservicos-siafi.tesouro.gov.br/siafi{ANO_SIAFI}he/services/cpr/manterContasPagarReceber?wsdl",
        "orcamentario": f"https://homextservicos-siafi.tesouro.gov.br/siafi{ANO_SIAFI}he/services/orcamentario/manterOrcamentario?wsdl",
    },
    "producao": {
        "contas_pagar_receber": f"https://servicos-siafi.tesouro.gov.br/siafi{ANO_SIAFI}/services/cpr/manterContasPagarReceber?wsdl",
        "orcamentario": f"https://servicos-siafi.tesouro.gov.br/siafi{ANO_SIAFI}/services/orcamentario/manterOrcamentario?wsdl",
    }
}

# ============================================
# TESTE 1: Verificar Tesouro Gerencial
# ============================================
def verificar_tesouro_gerencial():
    """Verifica acesso ao Tesouro Gerencial"""
    print("\n" + "="*60)
    print("TESTE 1: VERIFICANDO ACESSO AO TESOURO GERENCIAL")
    print("="*60)
    
    url_tg = "https://tesourogerencial.tesouro.gov.br"
    
    try:
        response = requests.get(url_tg, timeout=10, verify=False)
        print(f"\n✓ Tesouro Gerencial acessível: {response.status_code}")
        
        if response.status_code == 200:
            print("  → Suas credenciais do TG podem ser usadas aqui")
            print("  → Mas para APIs, você precisa de certificado digital")
        
        return True
    except Exception as e:
        print(f"✗ Erro ao acessar: {str(e)}")
        return False


# ============================================
# TESTE 2: Testar WSDL (SEM certificado)
# ============================================
def testar_wsdl_disponibilidade():
    """Testa se consegue acessar o WSDL (requer certificado)"""
    print("\n" + "="*60)
    print("TESTE 2: VERIFICANDO DISPONIBILIDADE DO WSDL")
    print("="*60)
    
    servico = "contas_pagar_receber"
    url_wsdl = WSDL_URLS[AMBIENTE][servico]
    
    print(f"\nAmbiente: {AMBIENTE.upper()}")
    print(f"Serviço: {servico}")
    print(f"URL: {url_wsdl}")
    
    try:
        # Tenta acessar sem certificado (vai falhar)
        response = requests.get(url_wsdl, timeout=10, verify=False)
        
        if response.status_code == 200:
            print("\n✓ WSDL acessível!")
            print("  → Você tem certificado configurado corretamente")
            return True
        else:
            print(f"\n⚠ Status: {response.status_code}")
            print("  → Você ainda não tem certificado configurado")
            return False
            
    except requests.exceptions.SSLError as e:
        print("\n⚠ ERRO DE CERTIFICADO SSL")
        print("  → Isso é ESPERADO se você não tem certificado digital")
        print("  → A API SOAP do SIAFI EXIGE certificado A1 de equipamento")
        return False
    except Exception as e:
        print(f"\n✗ ERRO: {str(e)}")
        return False


# ============================================
# TESTE 3: Informações sobre requisitos
# ============================================
def mostrar_requisitos():
    """Mostra os requisitos para acessar a API"""
    print("\n" + "="*60)
    print("REQUISITOS PARA ACESSAR API SOAP DO SIAFI")
    print("="*60)
    
    print("\n📋 VOCÊ PRECISA DE:")
    print("\n1. CERTIFICADO DIGITAL A1 DE EQUIPAMENTO")
    print("   • Tipo: A1 para equipamento (NÃO e-CPF ou e-CNPJ)")
    print("   • Emissor: Autoridade certificadora ICP-Brasil")
    print("   • Formato: .pfx (PKCS #12)")
    
    print("\n2. CADASTRO DO SISTEMA NO TESOURO NACIONAL")
    print("   • Abrir chamado no e-Serviços do Tesouro")
    print("   • Categoria: Universo SIAFI → SIAFI → WEBSERVICES")
    print("   • URL: https://e-servicos.tesouro.gov.br")
    
    print("\n3. USUÁRIOS CADASTRADOS NO SIAFI")
    print("   • Mesmos usuários do Tesouro Gerencial")
    print("   • Perfis de acesso adequados às operações")
    
    print("\n📝 IMPORTANTE:")
    print("   • Comece SEMPRE no ambiente de HOMOLOGAÇÃO")
    print("   • Só vá para PRODUÇÃO após testes completos")
    print("   • API funciona apenas em horário comercial")


# ============================================
# TESTE 4: Como prosseguir
# ============================================
def proximos_passos():
    """Mostra os próximos passos"""
    print("\n" + "="*60)
    print("PRÓXIMOS PASSOS")
    print("="*60)
    
    print("\n🔹 PASSO 1: Obter Certificado Digital")
    print("   1. Procure uma Autoridade Certificadora ICP-Brasil")
    print("   2. Solicite um certificado A1 de EQUIPAMENTO")
    print("   3. Guarde o arquivo .pfx com segurança")
    
    print("\n🔹 PASSO 2: Cadastrar Sistema no Tesouro")
    print("   1. Acesse: https://e-servicos.tesouro.gov.br")
    print("   2. Abra chamado em: Universo SIAFI → SIAFI → WEBSERVICES")
    print("   3. Informe:")
    print("      • Nome do sistema")
    print("      • Certificado digital (.cer)")
    print("      • Serviços que vai usar")
    
    print("\n🔹 PASSO 3: Testar em Homologação")
    print("   1. Configure certificado no código Python")
    print("   2. Use ambiente de homologação")
    print("   3. Teste todas as operações necessárias")
    
    print("\n🔹 PASSO 4: Usar em Produção")
    print("   1. Após aprovação dos testes")
    print("   2. Solicite migração para produção")
    print("   3. Configure ambiente de produção no código")
    
    print("\n📞 SUPORTE:")
    print("   • E-mail: ccont@tesouro.gov.br")
    print("   • Telefone: (61) 3412-3805")
    print("   • Horário: Comercial")


# ============================================
# TESTE 5: Consultar contratos (EXEMPLO)
# ============================================
def exemplo_com_certificado():
    """Exemplo de como usar com certificado (não funciona sem)"""
    print("\n" + "="*60)
    print("EXEMPLO DE CÓDIGO COM CERTIFICADO")
    print("="*60)
    
    codigo_exemplo = '''
# Depois de obter o certificado, use assim:

from zeep import Client
from zeep.transports import Transport
import requests

# Configurar certificado
cert_path = '/caminho/para/certificado.pfx'
cert_password = 'senha_do_certificado'

# Criar sessão com certificado
session = requests.Session()
session.cert = (cert_path, cert_password)
session.verify = False  # Ou caminho para CA bundle

# Criar transporte
transport = Transport(session=session)

# URL do WSDL
url_wsdl = "https://homextservicos-siafi.tesouro.gov.br/siafi2026he/services/cpr/manterContasPagarReceber?wsdl"

# Criar cliente SOAP
client = Client(url_wsdl, transport=transport)

# Ver operações disponíveis
print("Operações disponíveis:")
for service in client.wsdl.services.values():
    for port in service.ports.values():
        operations = port.binding._operations.values()
        for operation in operations:
            print(f"  - {operation.name}")

# Exemplo de chamada (ajuste conforme documentação)
resultado = client.service.ConsultarEmpenho(
    cpf=USUARIO,
    orgao='20401',  # Código da Justiça Federal CE
    ano='2026',
    numero_empenho='123456'
)
'''
    
    print(codigo_exemplo)


# ============================================
# EXECUTAR TODOS OS TESTES
# ============================================
if __name__ == "__main__":
    print("="*60)
    print("TESTE DE ACESSO - API SOAP SIAFI")
    print("Tesouro Nacional")
    print("="*60)
    
    # Verifica se configurou as credenciais
    if USUARIO == "seu_usuario":
        print("\n⚠️  Configure suas credenciais no código!")
        print("   Edite USUARIO e SENHA no início do arquivo")
        print("\n   (Mas saiba que apenas as credenciais NÃO bastam,")
        print("    você precisa de certificado digital)")
    
    # Executa testes
    verificar_tesouro_gerencial()
    testar_wsdl_disponibilidade()
    mostrar_requisitos()
    proximos_passos()
    exemplo_com_certificado()
    
    print("\n" + "="*60)
    print("CONCLUSÃO")
    print("="*60)
    print("\n✓ Você TEM acesso ao Tesouro Gerencial")
    print("✗ Você NÃO TEM certificado digital configurado")
    print("\n→ Siga os PRÓXIMOS PASSOS acima para obter acesso completo")
    print("\n💡 Dica: O chamado CH202600906 que você abriu é o lugar")
    print("   certo para tirar dúvidas sobre o processo!")
```
