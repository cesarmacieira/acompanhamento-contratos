```python
import pandas as pd
from pathlib import Path

# ==============================
# CONFIGURAÇÃO
# ==============================

pasta = Path(r"C:\Users\Cesar\Desktop\PNUD\acompanhamento-contratos")

arquivos = [
    "Dados 2023.xlsx",
    "Dados 2024.xlsx",
    "Dados 2025.xlsx",
    "Dados 2026.xlsx"
]

# ==============================
# PROCESSAMENTO
# ==============================

for nome in arquivos:
    caminho_excel = pasta / nome
    caminho_parquet = pasta / nome.replace(".xlsx", ".parquet")

    print(f"\n▶ Lendo {nome}...")

    # 1️⃣ Lê tudo como string (evita erro de tipo)
    df = pd.read_excel(
        caminho_excel,
        engine="openpyxl",
        dtype=str
    )

    print(f"✔ Arquivo lido | linhas: {len(df):,} | colunas: {df.shape[1]}")

    # 2️⃣ Converte automaticamente colunas numéricas válidas
    for col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="ignore")

    # 3️⃣ Salva em Parquet
    print(f"💾 Salvando {caminho_parquet.name}...")

    df.to_parquet(
        caminho_parquet,
        engine="pyarrow",
        compression="snappy",
        index=False
    )

    print("✅ Concluído")

print("\n🎉 Conversão finalizada com sucesso!")
```
